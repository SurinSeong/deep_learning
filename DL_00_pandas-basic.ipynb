## 새로운 열 만들기

```python
# 멀티 인덱스 생성
df = \
pd.DataFrame(
    {
        'a' : [4, 5, 6, 7],
        'b' : [7, 8, 9, 10],
        'c' : [12, 13, 15, 16]
    },
    index=pd.MultiIndex.from_tuples(
        [('d', 1), ('d', 2), ('e', 1), ('e', 2)]
    )
)
df
```

```python
# 조건에 맞는 새로운 열 만들기
df['Sum'] = df.a + df.b + df.c
df
```

```python
# assign() 이용, 조건에 맞는 새로운 열 만들기
# a, b, c열 값을 모두 더해서 d열 생성하기
df = df.assign(multiply=lambda df : df.a * df.b * df.c)
df
```

```python
# 숫자형 데이터를 구간 나누기
# a열을 두개로 나누어 각각 새롭게 레이블을 만드는 방법
df['Qcut'] = pd.qcut(df.a, 2, labels=['600 이하', '600 이상'])
df
```

```python
# 기준 값 이하와 이상을 모두 통일 시키기
# a열에서 5 이하는 모두 5로, 6 이상은 모두 6으로 변환
df['Clip'] = df.a.clip(lower=5, upper=6)
df
```

```python
# 최댓값 불러오기
df.max(axis=0) # 서로 다른 행과 행을 비교하기 (열 방향)
# df.max(axis=1) : 행 방향
```

```python
df.drop(columns='Qcut').max(axis=1)
```

```python
df.min(axis=0)
```

```python
# 행과 열 변환
df = \
pd.DataFrame({'A' : {0:'a', 1:'b', 2:'c'},
              'B' : {0:1, 1:3, 2:5},
              'C' : {0:2, 1:4, 2:6}
             })
df
```

* melt ==> 연습하기

```python
df.melt()
```

```python
# 하나의 열만 행으로 이동시키기
pd.melt(df, id_vars=['A'], value_vars=['B']) # A열만 그대로, B열은 행으로 이동
```

```python
# 여러 개 열을 행으로 이동시키기
# A열은 고정, 나머지 B와 C열은 행으로 이동
df_melt = df.melt(id_vars=['A'], value_vars=['B', 'C'])
df_melt
```

* pivot 연습 !

```python
# 특정 열의 값을 기준, 새로운 열 만들기
df_pivot = df_melt.pivot(index='A', columns='variable', values='value')
df_pivot
```

```python
# 원래의 데이터 형태로 복원하기
df_origin = df_pivot.reset_index()
df_origin.columns.name = None
df_origin
```

## 데이터프레임 연결하기
* merge(), concat()

```python
df1 = \
pd.DataFrame([['a', 1],
              ['b', 2]],
             columns=['letter', 'number'])

df2 = \
pd.DataFrame([['c', 3],
              ['e', 4]],
             columns=['letter', 'number'])
```

```python
df3 = \
pd.DataFrame([['c', 3, 'cat'],
              ['d', 4, 'dog']],
             columns=['letter', 'number', 'animal'])

df4 = \
pd.DataFrame([['bird', 'polly'],
              ['monkey', 'george']],
             columns=['animal', 'name'])
```

```python
# 데이터 프레임 합치기
pd.concat([df1, df2])
```

```python
pd.concat([df1, df3]) # axis=0 (default)
```

```python
pd.concat([df1, df3], axis=1)
```

```python
# 함께 공유하는 열만 합치기
pd.concat([df1, df3], join='inner')
```

```python
# merge()
pd.merge(df2, df3, how='left', on='number') # 왼쪽 df로
```

```python
pd.merge(df2, df3, how='right', on='number') # 오른쪽 df로
```

```python
# outer
pd.merge(df2, df3, how='outer', on='number')
```

```python
# 특정 열 비교, 공통값이 존재하는 경우만 추출
df2[df2.letter.isin(df3.letter)]
```

```python
df2[~df2.letter.isin(df3.letter)]
```

```python
# 공통값이 있는 것만 병합
pd.merge(df2, df3)
```

```python
# 어디서 병합되었는지 source 표시하기
pd.merge(df2, df3, how='outer', indicator=True)
```

```python
# 원하는 병합만 남기기
pd.merge(df2, df3, how='outer', indicator=True).query('_merge == "left_only"')
```

```python
# merge 컬럼 없애기
pd.merge(df2, df3, how='outer', indicator=True).query('_merge == "left_only"').drop(columns=['_merge'])
```
